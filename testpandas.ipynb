```python
import pandas as pd 
import numpy as np

df = pd.read_csv("datasets/train.csv")
```

```python
def replace_median_ages(df: pd.DataFrame) -> float:
    median_value = df["Age"].median()
    seriees_obj = df["Age"].fillna(value = median_value)
    df["Age"] = seriees_obj
    return df
```

```python
df =replace_median_ages(df)
```

```python
def sur_in_group(df, feature, value):
    col = df[df[feature] == value]
    
    return len(col), sum(col["Survived"])
```

```python
sur_in_group(df, "Pclass", 1)
```

```python
def empty_value_counter(df, feature):
    return sum(df[feature].isna())
```

```python
def group_year_range(df: pd.DataFrame, step_year : int = 10) -> dict:
    from collections import defaultdict
    
    seriees_obj = (df["Age"] // step_year)*step_year
    return defaultdict(int,seriees_obj.fillna(-1).value_counts(dropna = False))
```

```python
group_year_range(df)
```

```python
def survived_in_range(dataset_df: pd.DataFrame, age_range):
    age_categories = pd.cut(dataset_df["Age"], bins = age_range, right = False)
    return age_categories
    
```

```python
survived_in_range(df, [0, 20, 100])
```
